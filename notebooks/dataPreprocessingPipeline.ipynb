```python
# merging and joining data 

df_customers = ({
    "customer_id" : [1, 2, 3, 4],
    "name " : ["palak", "purvi " , "sakshi" , "hina" ]
})

df_order= ({
    "order_id" :[1020,102,105,106],
    "customer_id" : [1, 5, 3, 2],
    "amount" : [ 150,2500,6320,4251]
})
df1 = pd.DataFrame(df_customers)
df2 = pd.DataFrame(df_order)

pd.merge(df1, df2, on = "customer_id") # common in both  - inner join
pd.merge(df1, df2, on = "customer_id" , how = "left") #left join = left df1
pd.merge(df1, df2, on = "customer_id" , how = "right")#right join = right df2
pd.merge(df1, df2, on = "customer_id" , how = "outer")#outer - both left and right
```

```python
# data concatenation  - either by row(in stack) or column(side by side)
df1 = pd.DataFrame({
    "id" : [1, 2, 3, 4],
    "name" :["mansi" , "rahul " , "raghav", "romeo"]
})

df2 = pd.DataFrame({
    "id" : [4, 5, 6, 7],
    "name" :[" romeo ","saves ","me ","wecanbe"]
})

pd.concat([df1,df2])
pd.concat([df1,df2], axis = 1) #col as side by side
    
```
